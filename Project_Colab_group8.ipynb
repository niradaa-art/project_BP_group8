## เตรียมข้อมูล

### ตาราง GPP เฉพาะปี พ.ศ.2566

```python
# เลือกเฉพาะปี 2566
gpp_2566 = gpp[gpp["TIME_PERIOD"] == 2566]
gpp_2566
```

### ตาราง งบประมาณ แสดงเฉพาะจังหวัด แผนงาน ประเภทงบประมาณ และ Budget(หน่วย : ล้านบาท)

```python
# โหลดข้อมูลต้นฉบับใหม่เพื่อนำคอลัมน์ที่ถูกลบกลับมา
budget_temp = pd.read_csv(f"{base_path}/provincial_budget.csv", sep="|")

# เลือกเฉพาะคอลัมน์ที่ต้องการ
budget_df = budget_temp[["changwad_area_name", "plan_name", "objc", "budget"]].copy()

# แปลง Budget เป็นตัวเลข และเปลี่ยนหน่วยเป็นล้านบาท
budget_df["budget"] = pd.to_numeric(budget_df["budget"], errors="coerce") / 1_000_000
budget_df
```

### ตาราง งบประมาณรวมแต่ละจังหวัด

```python
# เลือกเฉพาะคอลัมน์ที่ต้องการ
budget = budget[["changwad_area_name", "budget"]]

# แปลง Budget เป็นตัวเลข
budget["budget"] = pd.to_numeric(budget["budget"], errors="coerce")

# รวมงบประมาณของแต่ละจังหวัด
budget_province = budget.groupby("changwad_area_name", as_index=False)["budget"].sum()
budget_province
```

```python
# ==========================================
# เตรียมข้อมูลและ Merge Datasets รวม
# ==========================================

# 1. จัดการข้อมูล GPP
def fix_province_name(name):
    if pd.isna(name):
        return name
    name = str(name).strip()
    if name == 'กรุงเทพมหานคร':
        return name
    elif not name.startswith('จังหวัด'):
        return f"จังหวัด{name}"
    return name

# กรองปี 2566 และตัดแถวที่เป็นภาพรวม ('รวม') ออก
gpp_2566_clean = gpp_2566[gpp_2566['CWT'] != 'รวม'].copy()

gpp_total = gpp_2566_clean[gpp_2566_clean['NI'] == 'ผลิตภัณฑ์มวลรวมจังหวัด (GPP)'][['CWT', 'Value']].rename(columns={'CWT': 'changwad_area_name', 'Value': 'GPP_Million'})
gpp_per_capita = gpp_2566_clean[gpp_2566_clean['NI'] == 'ผลิตภัณฑ์จังหวัด (GPP) ต่อหัว'][['CWT', 'Value']].rename(columns={'CWT': 'changwad_area_name', 'Value': 'GPP_Per_Capita'})

# ปรับชื่อจังหวัดใน GPP
gpp_total['changwad_area_name'] = gpp_total['changwad_area_name'].apply(fix_province_name)
gpp_per_capita['changwad_area_name'] = gpp_per_capita['changwad_area_name'].apply(fix_province_name)

# 2. จัดการข้อมูลพรรคการเมือง (ลบช่องว่างส่วนเกิน)
party_clean = party.copy()
party_clean['changwad_area_name'] = party_clean['changwad_area_name'].astype(str).str.strip()
party_clean['party_name'] = party_clean['party_name'].astype(str).str.strip()

# 3. รวมงบประมาณแยกตามจังหวัด
budget_prov_sum = budget_df.groupby('changwad_area_name', as_index=False)['budget'].sum()
budget_prov_sum['changwad_area_name'] = budget_prov_sum['changwad_area_name'].astype(str).str.strip()

# 4. Merge ข้อมูลเข้าด้วยกัน
merged_df = budget_prov_sum.merge(gpp_total, on='changwad_area_name', how='left')
merged_df = merged_df.merge(gpp_per_capita, on='changwad_area_name', how='left')
merged_df = merged_df.merge(party_clean, on='changwad_area_name', how='left')

# คัดกรองเฉพาะ 77 จังหวัด (ตัดส่วนกลางออก)
df_prov_only = merged_df[merged_df['changwad_area_name'] != 'ส่วนกลาง'].dropna(subset=['GPP_Million']).reset_index(drop=True)
```

```python
# แสดง 5 แถวแรกของตารางข้อมูลรวมระดับจังหวัด
df_prov_only.head()
```
