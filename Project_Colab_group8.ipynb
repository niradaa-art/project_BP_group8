```python
# ==========================================
# เตรียมข้อมูลและ Merge Datasets รวม
# ==========================================

# 1. จัดการข้อมูล GPP
def fix_province_name(name):
    if pd.isna(name):
        return name
    name = str(name).strip()
    if name == 'กรุงเทพมหานคร':
        return name
    elif not name.startswith('จังหวัด'):
        return f"จังหวัด{name}"
    return name

# กรองปี 2566 และตัดแถวที่เป็นภาพรวม ('รวม') ออก
gpp_2566_clean = gpp_2566[gpp_2566['CWT'] != 'รวม'].copy()

gpp_total = gpp_2566_clean[gpp_2566_clean['NI'] == 'ผลิตภัณฑ์มวลรวมจังหวัด (GPP)'][['CWT', 'Value']].rename(columns={'CWT': 'changwad_area_name', 'Value': 'GPP_Million'})
gpp_per_capita = gpp_2566_clean[gpp_2566_clean['NI'] == 'ผลิตภัณฑ์จังหวัด (GPP) ต่อหัว'][['CWT', 'Value']].rename(columns={'CWT': 'changwad_area_name', 'Value': 'GPP_Per_Capita'})

# ปรับชื่อจังหวัดใน GPP
gpp_total['changwad_area_name'] = gpp_total['changwad_area_name'].apply(fix_province_name)
gpp_per_capita['changwad_area_name'] = gpp_per_capita['changwad_area_name'].apply(fix_province_name)

# 2. จัดการข้อมูลพรรคการเมือง (ลบช่องว่างส่วนเกิน)
party_clean = party.copy()
party_clean['changwad_area_name'] = party_clean['changwad_area_name'].astype(str).str.strip()
party_clean['party_name'] = party_clean['party_name'].astype(str).str.strip()

# 3. รวมงบประมาณแยกตามจังหวัด
budget_prov_sum = budget_df.groupby('changwad_area_name', as_index=False)['budget'].sum()
budget_prov_sum['changwad_area_name'] = budget_prov_sum['changwad_area_name'].astype(str).str.strip()

# 4. Merge ข้อมูลเข้าด้วยกัน
merged_df = budget_prov_sum.merge(gpp_total, on='changwad_area_name', how='left')
merged_df = merged_df.merge(gpp_per_capita, on='changwad_area_name', how='left')
merged_df = merged_df.merge(party_clean, on='changwad_area_name', how='left')

# คัดกรองเฉพาะ 77 จังหวัด (ตัดส่วนกลางออก)
df_prov_only = merged_df[merged_df['changwad_area_name'] != 'ส่วนกลาง'].dropna(subset=['GPP_Million']).reset_index(drop=True)
```

##วิเคราะห์ข้อมูล (Data Analysis)

```python

# คำถามที่ 1: วิเคราะห์ความสัมพันธ์ระหว่างงบประมาณกับ GPP จังหวัด ปี 2566
# รวมการเตรียมข้อมูล จัดอันดับ กราฟ และ Markdown สรุปผลไว้ใน Cell เดียว

import pandas as pd
import numpy as np
import plotly.express as px
from IPython.display import display, Markdown

# 1. ตรวจสอบและเตรียมข้อมูล
b = budget[["changwad_area_name", "budget"]].copy()
g = gpp.copy()

b["budget"] = pd.to_numeric(b["budget"], errors="coerce")

# ใช้เฉพาะงบที่ระบุพื้นที่จังหวัด ไม่รวมรายการ "ส่วนกลาง"
# หมายเหตุ: ต้องตรวจสอบความหมายของงบกรุงเทพมหานครเพิ่มเติม
b = b[
    b["changwad_area_name"].notna()
    & (b["changwad_area_name"] != "ส่วนกลาง")
].copy()

# งบประมาณต้นทางเป็นบาท แปลงเป็นล้านบาท
b["budget_million"] = b["budget"] / 1_000_000

# รวมงบประมาณรายจังหวัด
b_province = (
    b.groupby("changwad_area_name", as_index=False)["budget_million"]
    .sum()
)

# เลือก GPP จังหวัด ปี 2566 เท่านั้น (ไม่ใช่ GPP ต่อหัว)
g = g[
    (g["TIME_PERIOD"] == 2566)
    & (g["NI"] == "ผลิตภัณฑ์มวลรวมจังหวัด (GPP)")
].copy()

g["Value"] = pd.to_numeric(g["Value"], errors="coerce")

# ตรวจว่ามี GPP มากกว่าหนึ่งค่าต่อจังหวัดหรือไม่
if g["CWT"].duplicated().any():
    raise ValueError(
        "พบข้อมูล GPP ซ้ำในจังหวัดเดียวกัน "
        "โปรดตรวจสอบข้อมูลก่อนนำมารวม"
    )

g_province = g[["CWT", "Value"]].rename(
    columns={"CWT": "province", "Value": "gpp_million"}
)

# ทำชื่อจังหวัดให้ตรงกัน เช่น จังหวัดขอนแก่น -> ขอนแก่น
b_province["province"] = (
    b_province["changwad_area_name"]
    .str.replace(r"^จังหวัด", "", regex=True)
)

# เชื่อมงบประมาณกับ GPP ตามชื่อจังหวัด
result = b_province.merge(
    g_province,
    on="province",
    how="inner",
    validate="one_to_one"
)

result = result.dropna(
    subset=["budget_million", "gpp_million"]
).copy()

if len(result) < 3:
    raise ValueError(
        "จังหวัดที่เชื่อมข้อมูลได้มีน้อยเกินไป "
        "โปรดตรวจชื่อจังหวัดระหว่างสองตาราง"
    )

# 2. จัดอันดับ
result["อันดับงบประมาณ"] = (
    result["budget_million"]
    .rank(method="min", ascending=False)
    .astype(int)
)
result["อันดับ GPP"] = (
    result["gpp_million"]
    .rank(method="min", ascending=False)
    .astype(int)
)

result = result.sort_values("budget_million", ascending=False)

print(f"จำนวนจังหวัดที่เชื่อมข้อมูลได้: {len(result)} จังหวัด")
print("อันดับจังหวัดตามงบประมาณ 10 อันดับแรก")
display(
    result[
        ["province", "budget_million", "gpp_million",
         "อันดับงบประมาณ", "อันดับ GPP"]
    ].head(10).rename(columns={
        "province": "จังหวัด",
        "budget_million": "งบประมาณ (ล้านบาท)",
        "gpp_million": "GPP (ล้านบาท)"
    })
)

# 3. กราฟ Scatter Plot พร้อมเส้นแนวโน้ม
# ใช้ฟอนต์ที่รองรับภาษาไทยเป็นตัวเลือกสำรอง
fig = px.scatter(
    result,
    x="budget_million",
    y="gpp_million",
    hover_name="province",
    title="ความสัมพันธ์ระหว่างงบประมาณกับ GPP รายจังหวัด ปี 2566",
    labels={
        "budget_million": "งบประมาณ (ล้านบาท)",
        "gpp_million": "ผลิตภัณฑ์มวลรวมจังหวัด GPP (ล้านบาท)",
        "province": "จังหวัด"
    },
    trendline="ols"
)

fig.update_layout(
    template="plotly_white",
    font=dict(
        family="Tahoma, Leelawadee UI, Noto Sans Thai, sans-serif",
        size=13
    ),
    title_x=0.5
)
fig.show()

# 4. คำนวณความสัมพันธ์
pearson = result["budget_million"].corr(
    result["gpp_million"], method="pearson"
)
spearman = result["budget_million"].corr(
    result["gpp_million"], method="spearman"
)

top_budget = result.iloc[0]
top_gpp = result.loc[result["gpp_million"].idxmax()]


```

```python

# ==================================================
# ข้อ 2: งบประมาณที่จัดสรรไปยังจังหวัดต่าง ๆ
# มุ่งเน้นที่แผนงานด้านใด?
# ==================================================

import pandas as pd
import plotly.express as px
from IPython.display import display, Markdown

# 1. เตรียมข้อมูล
plan_df = budget[
    ["changwad_area_name", "plan_name", "budget"]
].copy()

plan_df["changwad_area_name"] = (
    plan_df["changwad_area_name"].astype("string").str.strip()
)
plan_df["plan_name"] = (
    plan_df["plan_name"].astype("string").str.strip()
)
plan_df["budget"] = pd.to_numeric(
    plan_df["budget"], errors="coerce"
)

# เลือกเฉพาะรายการระดับจังหวัด ไม่รวมส่วนกลาง
plan_df = plan_df[
    plan_df["changwad_area_name"]
    .str.startswith("จังหวัด", na=False)
].copy()

plan_df = plan_df.dropna(subset=["plan_name", "budget"])
plan_df = plan_df[
    plan_df["plan_name"].ne("")
    & plan_df["plan_name"].ne("<NA>")
]

# 2. รวมงบประมาณตามแผนงาน
plan_summary = (
    plan_df.groupby("plan_name", as_index=False)
    .agg(
        budget_total=("budget", "sum"),
        record_count=("budget", "size")
    )
)

# แปลงเป็นล้านบาท
plan_summary["budget_million"] = (
    plan_summary["budget_total"] / 1_000_000
)

# คำนวณสัดส่วนงบประมาณ
total_budget = plan_summary["budget_total"].sum()

plan_summary["share_percent"] = (
    plan_summary["budget_total"] / total_budget * 100
)

# 3. จัดอันดับงบประมาณจากมากไปน้อย
plan_summary = plan_summary.sort_values(
    "budget_total", ascending=False
).reset_index(drop=True)

plan_summary["rank"] = range(1, len(plan_summary) + 1)

# 4. แสดงตาราง 15 อันดับแรก
top15 = plan_summary.head(15).copy()

display(
    top15[
        ["rank", "plan_name", "budget_million",
         "share_percent", "record_count"]
    ].rename(columns={
        "rank": "อันดับ",
        "plan_name": "แผนงาน",
        "budget_million": "งบประมาณ (ล้านบาท)",
        "share_percent": "สัดส่วน (%)",
        "record_count": "จำนวนรายการ"
    }).round({
        "งบประมาณ (ล้านบาท)": 2,
        "สัดส่วน (%)": 2
    })
)

# 5. สร้างกราฟแท่ง 15 แผนงานที่ได้รับงบประมาณสูงสุด
chart_df = top15.sort_values(
    "budget_million", ascending=True
)

fig = px.bar(
    chart_df,
    x="budget_million",
    y="plan_name",
    orientation="h",
    text="share_percent",
    title="ข้อ 2: 15 อันดับแผนงานที่ได้รับงบประมาณสูงสุด",
    labels={
        "budget_million": "งบประมาณ (ล้านบาท)",
        "plan_name": "แผนงาน",
        "share_percent": "สัดส่วนงบประมาณ (%)"
    }
)

fig.update_traces(
    texttemplate="%{text:.2f}%",
    textposition="outside",
    cliponaxis=False
)

fig.update_layout(
    template="plotly_white",
    font=dict(
        family="Tahoma, Leelawadee UI, Noto Sans Thai, sans-serif",
        size=12
    ),
    title_x=0.5,
    height=700,
    margin=dict(l=30, r=90, t=80, b=50),
    yaxis=dict(title="", automargin=True),
    xaxis=dict(title="งบประมาณ (ล้านบาท)")
)

fig.show()

# 6. สรุปผลจากข้อมูลจริง
if len(plan_summary) > 0:
    top = plan_summary.iloc[0]


```

เมื่อพิจารณาจังหวัดที่ได้รับงบประมาณสูงสุด 5 อันดับแรก ได้แก่ นนทบุรี นครปฐม นครราชสีมา ชลบุรี และเชียงใหม่ พบว่ามีสัดส่วนงบประมาณรวมกันร้อยละ 38.22 ส่วนจังหวัดที่ได้รับงบประมาณสูงสุด 10 อันดับแรกมีสัดส่วนรวมร้อยละ 49.58 ของงบประมาณทั้งหมด สะท้อนให้เห็นว่างบประมาณเกือบครึ่งหนึ่งกระจุกตัวอยู่ในจังหวัดเพียง 10 จังหวัดจากทั้งหมด 76 จังหวัดที่นำมาวิเคราะห์

ข้อค้นพบที่สำคัญคือ จังหวัดนนทบุรีมีสัดส่วนงบประมาณสูงกว่าจังหวัดอันดับรองลงมาอย่างมาก จึงควรตรวจสอบเพิ่มเติมว่างบประมาณส่วนนี้เกี่ยวข้องกับหน่วยงานภาครัฐ โครงการขนาดใหญ่ หรือรายการงบประมาณเฉพาะประเภทใด ทั้งนี้ การที่จังหวัดหนึ่งได้รับงบประมาณสูงไม่ได้หมายความว่าประชาชนในจังหวัดนั้นได้รับประโยชน์มากกว่าจังหวัดอื่นเสมอไป เนื่องจากงบประมาณอาจเป็นงบของหน่วยงานที่ตั้งอยู่ในจังหวัด แต่ให้บริการหรือดำเนินงานครอบคลุมหลายพื้นที่

โดยสรุป ผลการวิเคราะห์เบื้องต้นชี้ให้เห็นถึงความไม่สม่ำเสมอในการกระจายงบประมาณระหว่างจังหวัด อย่างไรก็ตาม การประเมินความเป็นธรรมในการจัดสรรงบประมาณควรพิจารณาปัจจัยเพิ่มเติม เช่น จำนวนประชากร ขนาดเศรษฐกิจ ความต้องการบริการสาธารณะ และพื้นที่ที่ได้รับประโยชน์จากโครงการ รวมถึงควรตรวจสอบข้อมูลของจังหวัดนนทบุรีเป็นพิเศษเพื่อทำความเข้าใจสาเหตุของงบประมาณที่สูงผิดจากจังหวัดอื่น

**ข้อจำกัดของการวิเคราะห์:** การคำนวณนี้พิจารณาเฉพาะรายการที่ระบุชื่อพื้นที่ขึ้นต้นด้วยคำว่า “จังหวัด” และไม่รวมกรุงเทพมหานครหรือรายการส่วนกลาง อีกทั้งเป็นการรวมงบประมาณจากทุกปีงบประมาณที่มีอยู่ในชุดข้อมูล หากต้องการสรุปเฉพาะปีใดปีหนึ่ง ควรกรองข้อมูลตามปีงบประมาณก่อนวิเคราะห์


```python

# ============================================================
# ข้อ 5: โครงสร้างและสัดส่วนประเภทของงบประมาณของแต่ละจังหวัดแตกต่างกันอย่างไร?
# ============================================================

import pandas as pd
import plotly.express as px
from IPython.display import display, Markdown

# 1. เตรียมข้อมูล
budget_type_df = budget[
    ["fy", "changwad_area_name", "objc", "budget"]
].copy()

budget_type_df["changwad_area_name"] = (
    budget_type_df["changwad_area_name"]
    .astype("string").str.strip()
)

budget_type_df["objc"] = (
    budget_type_df["objc"]
    .astype("string").str.strip()
)

budget_type_df["budget"] = pd.to_numeric(
    budget_type_df["budget"], errors="coerce"
)

budget_type_df = budget_type_df.dropna(
    subset=["changwad_area_name", "objc", "budget"]
)

# เลือกเฉพาะรายการที่ขึ้นต้นด้วย "จังหวัด"
# ไม่รวมกรุงเทพมหานครและรายการส่วนกลาง
budget_type_df = budget_type_df[
    budget_type_df["changwad_area_name"]
    .str.startswith("จังหวัด", na=False)
].copy()

# ทำชื่อจังหวัดให้อยู่ในรูปแบบเดียวกัน
budget_type_df["province"] = (
    budget_type_df["changwad_area_name"]
    .str.replace(r"^จังหวัด\s*", "", regex=True)
    .str.strip()
)

# 2. รวมงบประมาณแยกตามจังหวัดและประเภท
province_type_summary = (
    budget_type_df.groupby(
        ["province", "objc"], as_index=False
    )
    .agg(budget_total=("budget", "sum"))
)

# 3. สร้างตาราง Pivot
budget_pivot = province_type_summary.pivot(
    index="province",
    columns="objc",
    values="budget_total"
).fillna(0)

# คำนวณงบประมาณรวมของแต่ละจังหวัด
budget_pivot["รวมงบประมาณ"] = budget_pivot.sum(axis=1)

# คำนวณสัดส่วนของงบประมาณแต่ละประเภทในแต่ละจังหวัด
type_columns = [
    col for col in budget_pivot.columns
    if col != "รวมงบประมาณ"
]

budget_share = budget_pivot[type_columns].div(
    budget_pivot["รวมงบประมาณ"], axis=0
) * 100

# เรียงจังหวัดตามงบประมาณรวมจากมากไปน้อย
budget_pivot = budget_pivot.sort_values(
    "รวมงบประมาณ", ascending=False
)

budget_share = budget_share.loc[budget_pivot.index]

# 4. แสดงตารางสัดส่วนงบประมาณของแต่ละจังหวัด
display(
    budget_share.round(2).rename_axis("จังหวัด").reset_index()
)

# 5. สรุปประเภทที่มีงบประมาณรวมสูงสุด
type_summary = (
    province_type_summary.groupby("objc", as_index=False)
    .agg(budget_total=("budget_total", "sum"))
    .sort_values("budget_total", ascending=False)
    .reset_index(drop=True)
)

total_budget = type_summary["budget_total"].sum()

type_summary["budget_million"] = (
    type_summary["budget_total"] / 1_000_000
)

type_summary["share_percent"] = (
    type_summary["budget_total"] / total_budget * 100
)

type_summary["rank"] = type_summary.index + 1

display(
    type_summary[
        ["rank", "objc", "budget_million", "share_percent"]
    ].rename(columns={
        "rank": "อันดับ",
        "objc": "ประเภทของงบประมาณ",
        "budget_million": "งบประมาณ (ล้านบาท)",
        "share_percent": "สัดส่วน (%)"
    }).round(2)
)

# 6. กราฟแท่งแสดงงบประมาณรวมตามประเภท
type_chart = type_summary.sort_values(
    "budget_million", ascending=True
)

fig1 = px.bar(
    type_chart,
    x="budget_million",
    y="objc",
    orientation="h",
    title="ข้อ 5: งบประมาณรวมจำแนกตามประเภท",
    labels={
        "budget_million": "งบประมาณ (ล้านบาท)",
        "objc": "ประเภทของงบประมาณ"
    },
    text="share_percent"
)

fig1.update_traces(
    texttemplate="%{text:.1f}%",
    textposition="outside",
    cliponaxis=False
)

fig1.update_layout(
    template="plotly_white",
    font=dict(
        family="Tahoma, Leelawadee UI, Noto Sans Thai, sans-serif",
        size=12
    ),
    title_x=0.5,
    height=max(450, 45 * len(type_chart)),
    yaxis=dict(title="", automargin=True),
    xaxis=dict(title="งบประมาณ (ล้านบาท)")
)

fig1.show()

# 7. Heatmap เปรียบเทียบสัดส่วนประเภทของงบประมาณรายจังหวัด
# เลือก 15 จังหวัดที่มีงบประมาณรวมสูงสุดเพื่อให้อ่านกราฟง่ายขึ้น
top15_provinces = budget_pivot.head(15).index

heatmap_df = budget_share.loc[top15_provinces, type_columns]

fig2 = px.imshow(
    heatmap_df,
    labels=dict(
        x="ประเภทของงบประมาณ",
        y="จังหวัด",
        color="สัดส่วน (%)"
    ),
    aspect="auto",
    color_continuous_scale=[
        [0.0, "#D6C5F0"],   # ค่าต่ำเป็นม่วงอ่อน ไม่ใช่สีขาว
        [0.25, "#A78BDA"],
        [0.5, "#8055BD"],
        [0.75, "#54278F"],
        [1.0, "#2D004B"]    # ค่าสูงเป็นม่วงเข้ม
    ],
    title="ข้อ 5: สัดส่วนประเภทของงบประมาณใน 15 จังหวัดที่ได้รับงบสูงสุด"
)

fig2.update_layout(
    template="plotly_white",
    font=dict(
        family="Tahoma, Leelawadee UI, Noto Sans Thai, sans-serif",
        size=11
    ),
    title_x=0.5,
    height=650,
    xaxis=dict(tickangle=-35)
)

fig2.show()

# 8. ประเภทงบประมาณหลักของแต่ละจังหวัด
dominant_type = budget_share.idxmax(axis=1).to_frame(
    name="ประเภทที่มีสัดส่วนสูงสุด"
)

dominant_type["สัดส่วนสูงสุด (%)"] = budget_share.max(axis=1)

dominant_type["งบประมาณรวม (ล้านบาท)"] = (
    budget_pivot["รวมงบประมาณ"] / 1_000_000
)

display(
    dominant_type.reset_index().rename(columns={
        "province": "จังหวัด"
    }).round(2)
)

# 9. สรุปผลเบื้องต้นจากข้อมูลจริง
top_type = type_summary.iloc[0]


```

จากการวิเคราะห์โครงสร้างงบประมาณของ 76 จังหวัด โดยจำแนกตามประเภทของงบประมาณ 5 ประเภท ได้แก่ งบลงทุน งบอุดหนุน งบบุคลากร งบดำเนินงาน และงบรายจ่ายอื่น พบว่างบประมาณแต่ละประเภทมีสัดส่วนแตกต่างกัน โดยงบลงทุนมีมูลค่าสูงสุด 369,340.44 ล้านบาท คิดเป็นร้อยละ 43.76 ของงบประมาณรวม รองลงมาคืองบอุดหนุน 315,613.06 ล้านบาท คิดเป็นร้อยละ 37.40 และงบบุคลากร 121,720.92 ล้านบาท คิดเป็นร้อยละ 14.42 ส่วนงบดำเนินงานและงบรายจ่ายอื่นมีสัดส่วนร้อยละ 3.05 และ 1.37 ตามลำดับ
