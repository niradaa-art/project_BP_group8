## EDA (Exploratory Data Analysis)

### EDA ขั้นที่ 1 — ตรวจสอบข้อมูลเบื้องต้น

```python
import pandas as pd
import numpy as np
from IPython.display import display

# ==========================================
# 1. Load all datasets using the correct base_path
# ==========================================

budget = pd.read_csv(
    f'{base_path}/provincial_budget.csv',
    sep='|'
)

gdp = pd.read_csv(
    f'{base_path}/gdp.csv'
)

party = pd.read_csv(
    f'{base_path}/partylist.csv'
)

datasets = {
    'Provincial Budget': budget,
    'GDP': gdp,
    'Party List': party
}

# ==========================================
# 2. Inspect dataset structure
# ==========================================

for name, data in datasets.items():
    print('\n' + '=' * 60)
    print(f'DATASET: {name}')
    print('=' * 60)

    print(f'Number of rows: {data.shape[0]:,}')
    print(f'Number of columns: {data.shape[1]}')

    print('\nColumn names:')
    print(data.columns.tolist())

    print('\nData types and non-null counts:')
    data.info()

    print('\nFirst 5 rows:')
    display(data.head())

# ==========================================
# 3. Check missing values
# ==========================================

for name, data in datasets.items():
    print(f'\n=== Missing values: {name} ===')

    missing = pd.DataFrame({
        'Missing_Count': data.isna().sum(),
        'Missing_Percent': (
            data.isna().mean() * 100
        ).round(2)
    })

    display(missing)

# ==========================================
# 4. Check duplicate rows
# ==========================================

for name, data in datasets.items():
    print(f'\n=== Duplicate rows: {name} ===')
    print('Duplicate rows:', data.duplicated().sum())

# ==========================================
# 5. Check unique values in key columns
# ==========================================

key_columns = {
    'Provincial Budget': [
        'fy', 'changwad_area_name',
        'min_name', 'plan_name', 'objc'
    ],
    'GDP': [
        'TIME_PERIOD', 'NA_STAT',
        'NI', 'AREA', 'CWT', 'UNIT', 'FREQ'
    ],
    'Party List': [
        'changwad_area_name', 'party_name'
    ]
}

for name, data in datasets.items():
    print(f'\n=== Unique values: {name} ===')

    for col in key_columns[name]:
        if col in data.columns:
            print(
                f'{col}: {data[col].nunique(dropna=True)} '
                'unique values'
            )

# ==========================================
# 6. Check numeric data
# ==========================================

for name, data in datasets.items():
    numeric = data.select_dtypes(include='number')

    print(f'\n=== Numeric summary: {name} ===')

    if not numeric.empty:
        display(numeric.describe().T)
    else:
        print('No numeric columns detected.')
```

เมื่อตรวจสอบข้อมูลซ้ำ พบว่าชุดข้อมูลงบประมาณมีแถวซ้ำ 1 แถว ขณะที่ข้อมูล GPP และ Party List ไม่พบแถวซ้ำ ข้อมูลงบประมาณครอบคลุมปีงบประมาณ พ.ศ. 2566 เพียงปีเดียว และมีค่าของพื้นที่ 78 ค่า ซึ่งรวมพื้นที่ส่วนกลาง ส่วนข้อมูล GPP ครอบคลุมปี พ.ศ. 2538–2566 และมีค่าตัวแปรหลายประเภท พื้นที่ และหน่วยวัด

### EDA ขั้นที่ 2 — ตรวจสอบคุณภาพข้อมูลและทำความสะอาดเบื้องต้น
ขั้นนี้ตรวจ Missing Values, แถวซ้ำ, ชนิดข้อมูล, ค่าตัวเลขผิดรูปแบบ และข้อความที่มีช่องว่าง โดยทำงานกับสำเนาเพื่อไม่แก้ทับตัวแปรดิบของผู้จัดทำ

```python
import matplotlib.pyplot as plt
# EDA 2: Quality checks and safe cleaning
budget_raw = pd.read_csv(f'{base_path}/provincial_budget.csv', sep='|')
gpp_raw = pd.read_csv(f'{base_path}/gdp.csv')
party_raw = pd.read_csv(f'{base_path}/partylist.csv')

for name, data in {'Budget': budget_raw, 'GPP': gpp_raw, 'Party List': party_raw}.items():
    print(f'\n{name}: shape={data.shape}, exact duplicate rows={data.duplicated().sum()}')
    display(pd.DataFrame({'dtype': data.dtypes.astype(str), 'missing_count': data.isna().sum(),
                          'missing_percent': (data.isna().mean()*100).round(2),
                          'unique_values': data.nunique(dropna=True)}))

budget_clean = budget_raw.copy()
gpp_clean = gpp_raw.copy()
party_clean = party_raw.copy()
budget_clean['budget'] = pd.to_numeric(budget_clean['budget'], errors='coerce')
gpp_clean['Value'] = pd.to_numeric(gpp_clean['Value'], errors='coerce')
print('Budget values not convertible to numeric:', int(budget_clean['budget'].isna().sum()))
print('GPP Value values not convertible to numeric:', int(gpp_clean['Value'].isna().sum()))
print('Negative budget rows:', int((budget_clean['budget'] < 0).sum()))
print('Zero budget rows:', int((budget_clean['budget'] == 0).sum()))
for col in ['changwad_area_name','min_name','agc_name','plan_name','output_name','objc']:
    if col in budget_clean and budget_clean[col].dtype == 'object':
        budget_clean[col] = budget_clean[col].astype('string').str.strip()
for col in ['CWT','NI','AREA','UNIT','FREQ']:
    if col in gpp_clean and gpp_clean[col].dtype == 'object':
        gpp_clean[col] = gpp_clean[col].astype('string').str.strip()
party_clean['changwad_area_name'] = party_clean['changwad_area_name'].astype('string').str.strip()
party_clean['party_name'] = party_clean['party_name'].astype('string').str.strip()
print('Party groups after trimming whitespace:', party_clean['party_name'].nunique(dropna=True))
print('Note: duplicate rows are flagged for review, not automatically deleted.')
```

## ข้อจำกัด (Limitations)

1. **จำนวนรายการไม่ใช่มูลค่างบประมาณ:** การจัดอันดับพื้นที่ หน่วยงาน แผนงาน และประเภทงบรายจ่ายในผลลัพธ์นี้ใช้จำนวนแถว ไม่ใช่ผลรวมของ `budget` จึงไม่สามารถใช้แทนการจัดอันดับตามวงเงินงบประมาณได้
2. **ข้อมูล GPP มีหลายตัวชี้วัดและหน่วยวัด:** สถิติเชิงพรรณนาของ `Value` คำนวณจากข้อมูลทุกประเภทและทุกปี จึงไม่เหมาะสำหรับตีความระดับ GPP รายจังหวัดโดยตรง
3. **จำนวนจังหวัดในแต่ละกลุ่มพรรคไม่เท่ากัน:** บางกลุ่มมีจังหวัดจำนวนมาก ขณะที่บางกลุ่มมีเพียง 1–2 จังหวัด ค่าเฉลี่ย GPP ของกลุ่มขนาดเล็กจึงอาจเปลี่ยนแปลงได้มากจากข้อมูลของจังหวัดเพียงแห่งเดียว
4. **ความหมายของข้อมูลพรรคการเมืองยังต้องตรวจสอบ:** ชุดข้อมูลระบุพรรคการเมืองตามจังหวัด แต่ผลที่ให้มายังไม่ได้อธิบายหลักเกณฑ์ในการกำหนดพรรคให้แต่ละจังหวัด จึงควรระมัดระวังในการตีความผลการเปรียบเทียบ
5. **ยังไม่มีผลการวิเคราะห์ความสัมพันธ์ในขั้นนี้:** ผลลัพธ์เป็นเพียงสถิติเชิงพรรณนา ไม่สามารถใช้ยืนยันความสัมพันธ์หรือเหตุและผลระหว่างงบประมาณ GPP และพรรคการเมืองได้

### EDA ขั้นที่ 4 — Bivariate Analysis (สำรวจความสัมพันธ์ระหว่างตัวแปรสองตัว)
รวมงบประมาณให้เป็นระดับจังหวัดก่อน แล้วเปรียบเทียบกับ GPP รวมจังหวัดปีล่าสุดที่มีข้อมูล เพื่อให้หน่วยวิเคราะห์สอดคล้องกัน

```python
# EDA 4: provincial budget vs latest-year GPP

def normalize_province(name):
    if pd.isna(name): return np.nan
    name = str(name).strip()
    if name in ['รวม','ส่วนกลาง','nan','']: return name
    if name == 'กรุงเทพมหานคร': return name
    return name if name.startswith('จังหวัด') else 'จังหวัด' + name

budget_eda = budget_clean.copy()
budget_eda['province'] = budget_eda['changwad_area_name'].map(normalize_province)
budget_prov_eda = (budget_eda[budget_eda['province'].ne('ส่วนกลาง')]
                   .groupby('province', as_index=False)['budget'].sum()
                   .rename(columns={'budget':'Budget_Baht'}))
gpp_total_eda = gpp_clean[(gpp_clean['NI'].astype(str).str.strip() == 'ผลิตภัณฑ์มวลรวมจังหวัด (GPP)') &
                          (gpp_clean['CWT'].astype(str).str.strip() != 'รวม')].copy()
gpp_total_eda['province'] = gpp_total_eda['CWT'].map(normalize_province)
latest_year_eda = int(pd.to_numeric(gpp_total_eda['TIME_PERIOD']).max())
gpp_latest_eda = gpp_total_eda[pd.to_numeric(gpp_total_eda['TIME_PERIOD']) == latest_year_eda][['province','Value']].copy()
gpp_latest_eda['Value'] = pd.to_numeric(gpp_latest_eda['Value'], errors='coerce')
gpp_latest_eda = gpp_latest_eda.rename(columns={'Value':'GPP_Million_Baht'})
bivariate_eda = budget_prov_eda.merge(gpp_latest_eda, on='province', how='inner', validate='one_to_one')
print('GPP latest year:', latest_year_eda, '| matched provinces:', len(bivariate_eda))
print('Pearson correlation:', round(bivariate_eda[['Budget_Baht','GPP_Million_Baht']].corr(method='pearson').iloc[0,1], 3))
print('Spearman correlation:', round(bivariate_eda[['Budget_Baht','GPP_Million_Baht']].corr(method='spearman').iloc[0,1], 3))
print('Pearson correlation excluding Bangkok:', round(bivariate_eda[bivariate_eda['province']!='กรุงเทพมหานคร'][['Budget_Baht','GPP_Million_Baht']].corr().iloc[0,1], 3))
plt.figure(figsize=(8,5)); plt.scatter(bivariate_eda['GPP_Million_Baht'], bivariate_eda['Budget_Baht']/1e6, alpha=.75)
plt.xlabel(f'Provincial GPP ({latest_year_eda}, Million Baht)'); plt.ylabel('Provincial Budget (Million Baht)')
plt.title('EDA 4: Provincial Budget vs GPP'); plt.tight_layout(); plt.show()
```

#### ผลการวิเคราะห์ (Results)

จากการวิเคราะห์ความสัมพันธ์ระหว่างงบประมาณรายจังหวัดกับผลิตภัณฑ์มวลรวมจังหวัด (GPP) ในปี พ.ศ. 2566 พบว่าข้อมูลสามารถจับคู่จังหวัดได้ 77 จังหวัด โดยค่าสหสัมพันธ์เพียร์สัน (Pearson correlation) เท่ากับ 0.958 แสดงถึงความสัมพันธ์เชิงเส้นทางบวกในระดับสูงมาก กล่าวคือ จังหวัดที่มี GPP สูงมีแนวโน้มที่จะมีงบประมาณรวมสูงด้วยเมื่อพิจารณาจากข้อมูลทั้ง 77 จังหวัด

เมื่อพิจารณาค่าสหสัมพันธ์สเปียร์แมน (Spearman correlation) ซึ่งวัดความสัมพันธ์จากลำดับของข้อมูล พบว่ามีค่าเท่ากับ 0.694 แสดงถึงความสัมพันธ์ทางบวกในระดับค่อนข้างสูง สะท้อนว่าจังหวัดที่มีลำดับ GPP สูงมักมีแนวโน้มที่จะมีลำดับงบประมาณสูงเช่นกัน แต่ความสัมพันธ์ด้านลำดับไม่สูงเท่าความสัมพันธ์เชิงเส้นที่วัดด้วย Pearson

อย่างไรก็ตาม เมื่อไม่นำกรุงเทพมหานครมาร่วมวิเคราะห์ ค่า Pearson correlation ลดลงจาก 0.958 เหลือ 0.244 แสดงว่าความสัมพันธ์เชิงเส้นระหว่างงบประมาณกับ GPP ของจังหวัดที่เหลือมีระดับค่อนข้างต่ำเมื่อเทียบกับผลที่รวมกรุงเทพมหานครไว้ ผลนี้ชี้ให้เห็นว่าความสัมพันธ์ที่สูงมากในภาพรวมอาจได้รับอิทธิพลอย่างมากจากกรุงเทพมหานคร ซึ่งมีขนาดเศรษฐกิจและงบประมาณแตกต่างจากจังหวัดอื่น

## สรุปผล (Conclusion)

ผลการวิเคราะห์พบว่างบประมาณรายจังหวัดและ GPP ปี พ.ศ. 2566 มีความสัมพันธ์เชิงบวกสูงมากเมื่อพิจารณาครบทั้ง 77 จังหวัด แต่เมื่อแยกกรุงเทพมหานครออก ความสัมพันธ์เชิงเส้นลดลงอย่างชัดเจน จึงควรระมัดระวังในการสรุปว่าจังหวัดที่มีเศรษฐกิจขนาดใหญ่จะได้รับงบประมาณสูงตามไปด้วยในทุกกรณี

ผลดังกล่าวแสดงถึงความสัมพันธ์ระหว่างตัวแปร ไม่ได้ยืนยันว่าการมี GPP สูงเป็นสาเหตุให้ได้รับงบประมาณสูง หรือว่างบประมาณเป็นสาเหตุให้ GPP สูง การวิเคราะห์เพิ่มเติมควรพิจารณาการกระจายตัวของข้อมูล ความแตกต่างระหว่างกรุงเทพมหานครกับจังหวัดอื่น และงบประมาณเมื่อเทียบกับขนาดเศรษฐกิจของแต่ละจังหวัด

## ข้อจำกัด (Limitations)

1. **อิทธิพลของกรุงเทพมหานคร:** ค่า Pearson ลดลงอย่างมากเมื่อไม่รวมกรุงเทพมหานคร แสดงว่าความสัมพันธ์โดยรวมไวต่อค่าของพื้นที่ที่มีขนาดใหญ่แตกต่างจากพื้นที่อื่น
2. **ความสัมพันธ์ไม่เท่ากับเหตุและผล:** ค่าสหสัมพันธ์ไม่สามารถยืนยันได้ว่างบประมาณส่งผลต่อ GPP หรือ GPP เป็นตัวกำหนดงบประมาณ
3. **ขนาดเศรษฐกิจและการจัดสรรงบประมาณมีหลายปัจจัย:** การวิเคราะห์นี้ยังไม่ได้ควบคุมปัจจัยอื่น เช่น จำนวนประชากร ภารกิจของหน่วยงาน หรือความต้องการด้านบริการสาธารณะ
4. **เป็นการเปรียบเทียบเพียงปีเดียว:** ผลการวิเคราะห์ใช้ข้อมูลปี พ.ศ. 2566 จึงยังไม่สามารถสรุปแนวโน้มความสัมพันธ์ในระยะยาวได้
5. **ค่าสหสัมพันธ์ขึ้นอยู่กับรูปแบบข้อมูล:** Pearson วัดความสัมพันธ์เชิงเส้น ส่วน Spearman วัดความสัมพันธ์จากลำดับ ทั้งสองค่าอาจแตกต่างกันได้เมื่อข้อมูลมีค่าที่โดดเด่นหรือความสัมพันธ์ไม่ได้เป็นเส้นตรง

### EDA ขั้นที่ 5 — ตรวจความเข้ากันได้ของคีย์ก่อน Merge

```python
# Load raw data independently so the analysis does not depend on earlier column selections
budget_analysis_raw = pd.read_csv(f'{base_path}/provincial_budget.csv', sep='|')
gpp_analysis_raw = pd.read_csv(f'{base_path}/gdp.csv')
party_analysis_raw = pd.read_csv(f'{base_path}/partylist.csv')
budget_analysis_raw['budget'] = pd.to_numeric(budget_analysis_raw['budget'],errors='coerce')
budget_analysis_raw['province'] = budget_analysis_raw['changwad_area_name'].map(normalize_province)
province_budget = (budget_analysis_raw[budget_analysis_raw['province'].ne('ส่วนกลาง')]
                   .groupby('province',as_index=False)['budget'].sum()
                   .rename(columns={'budget':'Budget_Baht'}))
province_budget['Share_pct'] = province_budget['Budget_Baht']/province_budget['Budget_Baht'].sum()*100
province_budget = province_budget.sort_values('Budget_Baht',ascending=False).reset_index(drop=True)
province_budget['Rank'] = np.arange(1,len(province_budget)+1)
province_budget['Cumulative_share_pct'] = province_budget['Share_pct'].cumsum()
print('Total provincial budget (excluding central category):',f"{province_budget['Budget_Baht'].sum()/1e6:,.2f} million baht")
print('Highest-budget provinces:'); display(province_budget.head(10)[['Rank','province','Budget_Baht','Share_pct','Cumulative_share_pct']].style.format({'Budget_Baht':'{:,.0f}','Share_pct':'{:.2f}%','Cumulative_share_pct':'{:.2f}%'}))
print('Lowest-budget provinces:'); display(province_budget.tail(10).sort_values('Budget_Baht')[['province','Budget_Baht','Share_pct']].style.format({'Budget_Baht':'{:,.0f}','Share_pct':'{:.3f}%'}))
print(f"Top 5 province share: {province_budget.head(5)['Share_pct'].sum():.2f}%")
print(f"Top 10 province share: {province_budget.head(10)['Share_pct'].sum():.2f}%")
p = province_budget.head(10).sort_values('Budget_Baht').copy(); p['Rank_Label'] = 'Rank '+p['Rank'].astype(str)
plt.figure(figsize=(9,5)); plt.barh(p['Rank_Label'],p['Budget_Baht']/1e6)
plt.xlabel('Budget (Million Baht)'); plt.ylabel('Province Rank'); plt.title('Phase 1: Top 10 Provinces by Budget'); plt.tight_layout(); plt.show()
plt.figure(figsize=(8,5)); plt.plot(province_budget['Rank'],province_budget['Cumulative_share_pct'],marker='.')
plt.xlabel('Province Rank (Descending Budget)'); plt.ylabel('Cumulative Share (%)'); plt.title('Phase 1: Cumulative Budget Concentration'); plt.grid(alpha=.25); plt.tight_layout(); plt.show()
```

### Phase 2 — Budget Deep Dive

```python
budget_detail = budget_analysis_raw.copy()
budget_detail['budget_million'] = budget_detail['budget']/1e6
phase2_summaries = {}
for col, label in [('min_name','Agency / Ministry'),('agc_name','Agency'),('plan_name','Plan'),('objc','Budget Type')]:
    if col not in budget_detail.columns: continue
    summary = budget_detail.groupby(col,dropna=False,as_index=False)['budget_million'].sum().sort_values('budget_million',ascending=False)
    summary['Share_pct'] = summary['budget_million']/budget_detail['budget_million'].sum()*100
    phase2_summaries[col] = summary
    print(f'Top categories by total budget: {label}')
    display(summary.head(10).style.format({'budget_million':'{:,.2f}','Share_pct':'{:.2f}%'}))
    plot = summary.head(10).sort_values('budget_million').copy()
    plot['Category_Rank'] = ['Category '+str(i) for i in range(len(plot)-len(plot)+1,len(plot)+1)]
    plt.figure(figsize=(9,5)); plt.barh(plot['Category_Rank'],plot['budget_million'])
    plt.xlabel('Budget (Million Baht)'); plt.ylabel('Category Rank'); plt.title(f'Phase 2: Top 10 Categories by {label}')
    plt.tight_layout(); plt.show()
print('Budget total used as denominator (all rows including central category):',f"{budget_detail['budget_million'].sum():,.2f} million baht")
```

### Phase 3 — Budget × GPP
**คำถาม:** จังหวัดที่มี GPP สูงได้รับงบสูงหรือไม่ และงบประมาณสูง/ต่ำเมื่อเทียบกับ GPP เป็นอย่างไร?

การวิเคราะห์นี้แบ่งเป็น 2 ส่วนย่อยภายใน Phase 3 ได้แก่ (1) Correlation Analysis เพื่อวิเคราะห์ความสัมพันธ์ระหว่างงบประมาณกับ GPP และตรวจผลเมื่อไม่รวมกรุงเทพมหานคร และ (2) Regression Analysis เพื่อประมาณงบประมาณจาก GPP ด้วย Linear Regression โดยผลจากโมเดลเป็นการประมาณเชิงสถิติ ไม่ใช่ข้อเสนอจัดสรรงบประมาณจริง

ใช้ scatter plot สำหรับตัวแปรเชิงปริมาณ และรายงาน Pearson กับ Spearman ประกอบกัน


```python
gpp_a = gpp_analysis_raw.copy()
gpp_a['Value'] = pd.to_numeric(gpp_a['Value'],errors='coerce')
gpp_a['TIME_PERIOD'] = pd.to_numeric(gpp_a['TIME_PERIOD'],errors='coerce')
gpp_total_a = gpp_a[(gpp_a['NI'].astype(str).str.strip()=='ผลิตภัณฑ์มวลรวมจังหวัด (GPP)') & (gpp_a['CWT'].astype(str).str.strip()!='รวม')].copy()
gpp_total_a['province'] = gpp_total_a['CWT'].map(normalize_province)
latest_year = int(gpp_total_a['TIME_PERIOD'].max()); previous_year = latest_year-1
pivot_gpp = gpp_total_a.pivot_table(index='province',columns='TIME_PERIOD',values='Value',aggfunc='first')
latest_col = pivot_gpp[latest_year].rename('GPP_Million_Baht')
prev_col = pivot_gpp[previous_year].rename('GPP_Previous_Million_Baht') if previous_year in pivot_gpp.columns else pd.Series(dtype=float,name='GPP_Previous_Million_Baht')
analysis_df = province_budget.merge(latest_col,on='province',how='inner',validate='one_to_one')
analysis_df = analysis_df.merge(prev_col,on='province',how='left',validate='one_to_one')
analysis_df['GPP_Growth_pct'] = (analysis_df['GPP_Million_Baht']/analysis_df['GPP_Previous_Million_Baht']-1)*100
analysis_df['Budget_per_GPP_pct'] = analysis_df['Budget_Baht']/(analysis_df['GPP_Million_Baht']*1e6)*100
print('GPP year:',latest_year,'| matched provinces:',len(analysis_df))
print('Pearson Budget-GPP:',round(analysis_df[['Budget_Baht','GPP_Million_Baht']].corr().iloc[0,1],3))
print('Spearman Budget-GPP:',round(analysis_df[['Budget_Baht','GPP_Million_Baht']].corr(method='spearman').iloc[0,1],3))
no_bkk = analysis_df[analysis_df['province']!='กรุงเทพมหานคร']
print('Pearson excluding Bangkok:',round(no_bkk[['Budget_Baht','GPP_Million_Baht']].corr().iloc[0,1],3))
print('Pearson Budget-GPP growth:',round(analysis_df[['Budget_Baht','GPP_Growth_pct']].corr().iloc[0,1],3))
print('Spearman Budget-GPP growth:',round(analysis_df[['Budget_Baht','GPP_Growth_pct']].corr(method='spearman').iloc[0,1],3))
print('Top GPP provinces:'); display(analysis_df.nlargest(10,'GPP_Million_Baht')[['province','Budget_Baht','GPP_Million_Baht','GPP_Growth_pct','Budget_per_GPP_pct']].style.format({'Budget_Baht':'{:,.0f}','GPP_Million_Baht':'{:,.2f}','GPP_Growth_pct':'{:.2f}%','Budget_per_GPP_pct':'{:.2f}%'}))
print('Highest budget-to-GPP ratios (descriptive only):'); display(analysis_df.nlargest(10,'Budget_per_GPP_pct')[['province','Budget_per_GPP_pct']].style.format({'Budget_per_GPP_pct':'{:.2f}%'}))
print('Lowest budget-to-GPP ratios (descriptive only):'); display(analysis_df.nsmallest(10,'Budget_per_GPP_pct')[['province','Budget_per_GPP_pct']].style.format({'Budget_per_GPP_pct':'{:.2f}%'}))
plt.figure(figsize=(8,5)); plt.scatter(analysis_df['GPP_Million_Baht'],analysis_df['Budget_Baht']/1e6,alpha=.75)
plt.xlabel(f'Provincial GPP ({latest_year}, Million Baht)'); plt.ylabel('Provincial Budget (Million Baht)'); plt.title('Phase 3: Provincial Budget vs GPP'); plt.tight_layout(); plt.show()
plt.figure(figsize=(8,5)); plt.scatter(analysis_df['GPP_Growth_pct'],analysis_df['Budget_Baht']/1e6,alpha=.75)
plt.xlabel(f'GPP Growth ({previous_year}-{latest_year}, %)'); plt.ylabel('Provincial Budget (Million Baht)'); plt.title('Phase 3: Provincial Budget vs GPP Growth'); plt.tight_layout(); plt.show()
```

### Phase 3.2 — Regression Analysis: การประมาณงบประมาณจาก GPP
ส่วนนี้เป็นการวิเคราะห์ต่อยอดภายใน **Phase 3 — Budget × GPP** ไม่ใช่ Phase เพิ่มเติม ใช้ Linear Regression เพื่อศึกษาว่า GPP สัมพันธ์กับงบประมาณอย่างไร และทดลองคำนวณค่าประมาณภายใต้ค่า GPP เป้าหมาย โดยไม่ตีความว่าเป็นงบประมาณที่เหมาะสมหรือจำเป็นจริง


```python

# ============================================================
# Phase 3.2: Regression Analysis & Budget Estimation Model
# ประมาณงบประมาณจาก GPP เพื่อประกอบการวิเคราะห์ Phase 3
# ============================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# ============================================================
# 0. LOAD AND PREPARE DATA
# ============================================================

# หากมี base_path จากเซลล์ก่อนหน้า จะใช้ค่านั้น
# หากไม่มี จะใช้ path เริ่มต้นนี้
if "base_path" not in globals():
    base_path = "/content/drive/MyDrive/datasets_budget"

budget_path = os.path.join(base_path, "provincial_budget.csv")
gpp_path = os.path.join(base_path, "gdp.csv")
party_path = os.path.join(base_path, "partylist.csv")

for path in [budget_path, gpp_path, party_path]:
    if not os.path.exists(path):
        raise FileNotFoundError(
            f"ไม่พบไฟล์: {path}\n"
            "กรุณาตรวจสอบว่า Google Drive เชื่อมต่อแล้ว "
            "และ base_path ถูกต้อง"
        )

# โหลดข้อมูล
budget_raw = pd.read_csv(budget_path, sep="|")
gpp_raw = pd.read_csv(gpp_path)
party_raw = pd.read_csv(party_path)

# ตรวจสอบคอลัมน์ที่จำเป็น
required_budget = {"changwad_area_name", "budget"}
required_gpp = {"TIME_PERIOD", "NI", "CWT", "Value"}
required_party = {"changwad_area_name", "party_name"}

if not required_budget.issubset(budget_raw.columns):
    raise ValueError(
        f"Budget ขาดคอลัมน์: {required_budget - set(budget_raw.columns)}"
    )

if not required_gpp.issubset(gpp_raw.columns):
    raise ValueError(
        f"GPP ขาดคอลัมน์: {required_gpp - set(gpp_raw.columns)}"
    )

if not required_party.issubset(party_raw.columns):
    raise ValueError(
        f"Party List ขาดคอลัมน์: {required_party - set(party_raw.columns)}"
    )


def fix_province_name(name):
    """ปรับรูปแบบชื่อจังหวัดให้ตรงกันระหว่างไฟล์"""
    if pd.isna(name):
        return np.nan

    name = str(name).strip()

    if name in ["", "nan", "None"]:
        return np.nan

    if name in ["กรุงเทพมหานคร", "ส่วนกลาง", "รวม"]:
        return name

    if not name.startswith("จังหวัด"):
        return "จังหวัด" + name

    return name


# -------------------------
# 0.1 Budget
# -------------------------

budget_df = budget_raw[
    ["changwad_area_name", "budget"]
].copy()

budget_df["changwad_area_name"] = (
    budget_df["changwad_area_name"].apply(fix_province_name)
)

budget_df["budget"] = pd.to_numeric(
    budget_df["budget"].astype(str).str.replace(",", "", regex=False),
    errors="coerce"
)

# ตัดแถวที่ไม่มีชื่อจังหวัดหรือไม่มีค่างบประมาณ
budget_df = budget_df.dropna(
    subset=["changwad_area_name", "budget"]
)

# แปลงหน่วยจากบาทเป็นล้านบาท
budget_df["budget"] = budget_df["budget"] / 1_000_000

# รวมงบประมาณรายจังหวัด
budget_prov_sum = (
    budget_df.groupby("changwad_area_name", as_index=False)["budget"]
    .sum()
)

# ไม่รวมหมวดส่วนกลางในการวิเคราะห์ระดับจังหวัด
budget_prov_sum = budget_prov_sum[
    budget_prov_sum["changwad_area_name"] != "ส่วนกลาง"
].copy()


# -------------------------
# 0.2 GPP: ปีล่าสุดที่ต้องการวิเคราะห์
# -------------------------

gpp_df = gpp_raw.copy()

gpp_df["TIME_PERIOD"] = pd.to_numeric(
    gpp_df["TIME_PERIOD"], errors="coerce"
)

gpp_df["Value"] = pd.to_numeric(
    gpp_df["Value"].astype(str).str.replace(",", "", regex=False),
    errors="coerce"
)

gpp_df["NI"] = gpp_df["NI"].astype(str).str.strip()
gpp_df["CWT"] = gpp_df["CWT"].astype(str).str.strip()

# ใช้ปีล่าสุดที่มีข้อมูล GPP รวมระดับจังหวัด
gpp_total_all = gpp_df[
    gpp_df["NI"] == "ผลิตภัณฑ์มวลรวมจังหวัด (GPP)"
].copy()

gpp_total_all = gpp_total_all[
    gpp_total_all["CWT"] != "รวม"
].copy()

if gpp_total_all.empty:
    raise ValueError(
        "ไม่พบข้อมูล NI = 'ผลิตภัณฑ์มวลรวมจังหวัด (GPP)' "
        "กรุณาตรวจสอบค่าจริงในคอลัมน์ NI"
    )

latest_year = int(gpp_total_all["TIME_PERIOD"].max())

gpp_latest = gpp_total_all[
    gpp_total_all["TIME_PERIOD"] == latest_year
].copy()

gpp_latest["changwad_area_name"] = (
    gpp_latest["CWT"].apply(fix_province_name)
)

gpp_latest = gpp_latest[
    ["changwad_area_name", "Value"]
].rename(columns={"Value": "GPP_Million"})

# ป้องกันการ merge แบบ many-to-many หากข้อมูลมีชื่อจังหวัดซ้ำ
if gpp_latest["changwad_area_name"].duplicated().any():
    duplicates = gpp_latest.loc[
        gpp_latest["changwad_area_name"].duplicated(keep=False),
        "changwad_area_name"
    ].tolist()

    raise ValueError(
        f"พบชื่อจังหวัดซ้ำใน GPP ปี {latest_year}: {duplicates}\n"
        "กรุณาตรวจสอบ NI, UNIT และระดับข้อมูลก่อนนำมารวม"
    )


# -------------------------
# 0.3 Party List
# -------------------------

party_clean = party_raw[
    ["changwad_area_name", "party_name"]
].copy()

party_clean["changwad_area_name"] = (
    party_clean["changwad_area_name"].apply(fix_province_name)
)

party_clean["party_name"] = (
    party_clean["party_name"].astype(str).str.strip()
)

if party_clean["changwad_area_name"].duplicated().any():
    raise ValueError(
        "พบชื่อจังหวัดซ้ำใน Party List กรุณาตรวจสอบก่อน merge"
    )


# -------------------------
# 0.4 Merge datasets
# -------------------------

df_prov_only = (
    budget_prov_sum
    .merge(
        gpp_latest,
        on="changwad_area_name",
        how="inner",
        validate="one_to_one"
    )
    .merge(
        party_clean,
        on="changwad_area_name",
        how="left",
        validate="one_to_one"
    )
)

# ใช้เฉพาะข้อมูลที่จำเป็นต่อแบบจำลอง
econ_budget = df_prov_only.dropna(
    subset=["budget", "GPP_Million"]
).copy()

# GPP ต้องเป็นค่าบวกสำหรับการวิเคราะห์นี้
econ_budget = econ_budget[
    (econ_budget["GPP_Million"] > 0)
    & (econ_budget["budget"] >= 0)
].reset_index(drop=True)

if len(econ_budget) < 3:
    raise ValueError(
        "จำนวนจังหวัดที่ใช้สร้างแบบจำลองน้อยเกินไป "
        "กรุณาตรวจสอบข้อมูลและการจับคู่ชื่อจังหวัด"
    )

print("=" * 65)
print("DATA PREPARATION SUMMARY")
print("=" * 65)
print(f"GPP year: {latest_year}")
print(f"Provinces included in model: {len(econ_budget)}")
print(f"Total budget (million baht): {econ_budget['budget'].sum():,.2f}")
print(f"Total GPP (million baht): {econ_budget['GPP_Million'].sum():,.2f}")
print(f"Missing budget: {econ_budget['budget'].isna().sum()}")
print(f"Missing GPP: {econ_budget['GPP_Million'].isna().sum()}")


# ============================================================
# 1. FIT LINEAR REGRESSION MODEL
# ============================================================

X = econ_budget[["GPP_Million"]]
y = econ_budget["budget"]

model = LinearRegression()
model.fit(X, y)

econ_budget["Predicted_Budget"] = model.predict(X)
econ_budget["Residual"] = (
    econ_budget["budget"] - econ_budget["Predicted_Budget"]
)

r2 = r2_score(y, econ_budget["Predicted_Budget"])
mae = mean_absolute_error(y, econ_budget["Predicted_Budget"])
rmse = np.sqrt(mean_squared_error(y, econ_budget["Predicted_Budget"]))

slope = model.coef_[0]
intercept = model.intercept_

print("\n" + "=" * 65)
print("BUDGET ESTIMATION MODEL")
print("=" * 65)
print(
    "Estimated budget (million baht) = "
    f"{intercept:,.4f} + ({slope:,.4f} × GPP)"
)
print(f"R-squared (in-sample): {r2:.4f}")
print(f"MAE (million baht): {mae:,.2f}")
print(f"RMSE (million baht): {rmse:,.2f}")

print(
    "\nหมายเหตุ: ค่าประเมินข้างต้นคำนวณจากข้อมูลชุดเดียวกับที่ใช้ฝึกโมเดล "
    "จึงเป็นผลประเมิน in-sample ไม่ใช่ผลทดสอบกับข้อมูลที่โมเดลไม่เคยเห็น"
)


# ============================================================
# 2. GRAPH 1: BUDGET VS GPP + REGRESSION LINE
# ============================================================

plt.figure(figsize=(10, 6))

plt.scatter(
    econ_budget["GPP_Million"],
    econ_budget["budget"],
    alpha=0.75,
    label="Provincial data"
)

x_line = np.linspace(
    econ_budget["GPP_Million"].min(),
    econ_budget["GPP_Million"].max(),
    200
)

y_line = model.predict(pd.DataFrame({"GPP_Million": x_line}))

plt.plot(
    x_line,
    y_line,
    linewidth=2,
    label="Linear regression"
)

# เน้นกรุงเทพมหานครหากมีข้อมูล
bkk = econ_budget[
    econ_budget["changwad_area_name"] == "กรุงเทพมหานคร"
]

if not bkk.empty:
    plt.scatter(
        bkk["GPP_Million"],
        bkk["budget"],
        s=100,
        marker="x",
        label="Bangkok"
    )

plt.title(f"Provincial Budget vs GPP ({latest_year})")
plt.xlabel("GPP (Million Baht)")
plt.ylabel("Budget (Million Baht)")
plt.legend()
plt.grid(True, alpha=0.25)
plt.tight_layout()
plt.show()


# ============================================================
# 3. GRAPH 2: ACTUAL BUDGET VS PREDICTED BUDGET
# ============================================================

plt.figure(figsize=(8, 7))

plt.scatter(
    econ_budget["budget"],
    econ_budget["Predicted_Budget"],
    alpha=0.75
)

min_val = min(
    econ_budget["budget"].min(),
    econ_budget["Predicted_Budget"].min()
)

max_val = max(
    econ_budget["budget"].max(),
    econ_budget["Predicted_Budget"].max()
)

plt.plot(
    [min_val, max_val],
    [min_val, max_val],
    linestyle="--",
    linewidth=2,
    label="Perfect prediction"
)

plt.title("Actual vs Predicted Provincial Budget")
plt.xlabel("Actual Budget (Million Baht)")
plt.ylabel("Predicted Budget (Million Baht)")
plt.legend()
plt.grid(True, alpha=0.25)
plt.tight_layout()
plt.show()


# ============================================================
# 4. GRAPH 3: TARGET GPP SCENARIOS
# ============================================================

# สร้างกรณีจำลอง GPP ตั้งแต่ค่าปัจจุบันถึงเติบโต 20%
gpp_min = econ_budget["GPP_Million"].min()
gpp_max = econ_budget["GPP_Million"].max()

target_gpp_values = np.linspace(
    gpp_min,
    gpp_max * 1.20,
    200
)

target_budget_values = model.predict(
    pd.DataFrame({"GPP_Million": target_gpp_values})
)

plt.figure(figsize=(10, 6))

plt.plot(
    target_gpp_values,
    target_budget_values,
    linewidth=2,
    label="Model-estimated budget"
)

plt.title("Estimated Budget Under Different GPP Targets")
plt.xlabel("Target GPP (Million Baht)")
plt.ylabel("Estimated Budget (Million Baht)")
plt.legend()
plt.grid(True, alpha=0.25)
plt.tight_layout()
plt.show()


# ============================================================
# 5. FUNCTION: ESTIMATE BUDGET FOR A TARGET GPP
# ============================================================

def calculate_required_budget(
    target_gpp_mb,
    current_gpp_mb=None,
    prov_name=""
):
    """
    ประมาณงบประมาณตามความสัมพันธ์ระหว่าง GPP กับงบประมาณในข้อมูลเดิม

    target_gpp_mb: เป้าหมาย GPP หน่วยล้านบาท
    current_gpp_mb: GPP ปัจจุบัน หน่วยล้านบาท (ถ้ามี)
    prov_name: ชื่อจังหวัด (ถ้ามี)
    """

    if not np.isfinite(target_gpp_mb) or target_gpp_mb <= 0:
        raise ValueError("target_gpp_mb ต้องเป็นตัวเลขที่มากกว่า 0")

    predicted_target_budget = float(
        model.predict(pd.DataFrame({
            "GPP_Million": [target_gpp_mb]
        }))[0]
    )

    print("\n" + "=" * 60)
    print("BUDGET ESTIMATION SCENARIO")
    print("=" * 60)

    if prov_name:
        print(f"Province: {prov_name}")

    print(f"Target GPP: {target_gpp_mb:,.2f} million baht")
    print(
        "Model-estimated budget at target GPP: "
        f"{predicted_target_budget:,.2f} million baht"
    )

    if predicted_target_budget < 0:
        print(
            "WARNING: The linear model predicts a negative budget. "
            "This is not practically meaningful and indicates that "
            "the estimate should not be used as a budget recommendation."
        )

    if current_gpp_mb is not None:
        if not np.isfinite(current_gpp_mb) or current_gpp_mb <= 0:
            raise ValueError("current_gpp_mb ต้องเป็นตัวเลขที่มากกว่า 0")

        growth_needed = (
            (target_gpp_mb - current_gpp_mb) / current_gpp_mb
        ) * 100

        estimated_current_budget = float(
            model.predict(pd.DataFrame({
                "GPP_Million": [current_gpp_mb]
            }))[0]
        )

        estimated_budget_difference = (
            predicted_target_budget - estimated_current_budget
        )

        print(f"Current GPP: {current_gpp_mb:,.2f} million baht")
        print(f"Required GPP change: {growth_needed:+.2f}%")
        print(
            "Model-estimated budget at current GPP: "
            f"{estimated_current_budget:,.2f} million baht"
        )
        print(
            "Difference between target and current model estimates: "
            f"{estimated_budget_difference:+,.2f} million baht"
        )

    print("-" * 60)
    print(
        "Caution: This is a model-based association, not proof that "
        "allocating this budget will achieve the target GPP."
    )

    return predicted_target_budget


# ============================================================
# 6. EXAMPLE 1: GENERIC GPP TARGET
# ============================================================

calculate_required_budget(
    target_gpp_mb=100_000
)


# ============================================================
# 7. EXAMPLE 2: CHIANG MAI TARGET GPP +10%
# ============================================================

sample_prov = econ_budget[
    econ_budget["changwad_area_name"].str.contains(
        "เชียงใหม่",
        na=False
    )
]

if not sample_prov.empty:
    cur_gpp = float(sample_prov["GPP_Million"].iloc[0])
    target_gpp = cur_gpp * 1.10

    calculate_required_budget(
        target_gpp_mb=target_gpp,
        current_gpp_mb=cur_gpp,
        prov_name="เชียงใหม่"
    )
else:
    print(
        "ไม่พบข้อมูลจังหวัดเชียงใหม่ใน econ_budget "
        "กรุณาตรวจสอบชื่อจังหวัดและผลการ merge"
    )


# ============================================================
# 8. OPTIONAL: PROVINCE-LEVEL MODEL RESULTS TABLE
# ============================================================

result_columns = [
    "changwad_area_name",
    "GPP_Million",
    "budget",
    "Predicted_Budget",
    "Residual"
]

province_model_results = (
    econ_budget[result_columns]
    .sort_values("GPP_Million", ascending=False)
    .reset_index(drop=True)
)

province_model_results.columns = [
    "Province",
    "GPP_Million_Baht",
    "Actual_Budget_Million_Baht",
    "Estimated_Budget_Million_Baht",
    "Residual_Million_Baht"
]

display(province_model_results.head(10))
```

## ข้อจำกัด (Limitations)

- ค่า R-squared เป็นผลจากข้อมูลชุดเดียวกับที่ใช้ฝึกโมเดล (in-sample) จึงยังไม่ยืนยันความแม่นยำกับข้อมูลใหม่ ควรประเมินด้วย holdout set หรือ cross-validation หากต้องการใช้เพื่อการพยากรณ์
- ค่า MAE และ RMSE ค่อนข้างสูงในหน่วยล้านบาท และความคลาดเคลื่อนอาจแตกต่างกันมากระหว่างจังหวัด
- ค่าประมาณงบประมาณติดลบเมื่อ GPP เท่ากับ 100,000 ล้านบาทแสดงว่าโมเดลไม่เหมาะกับทุกช่วงของ GPP และไม่ควรนำสมการไปใช้โดยไม่ตรวจสอบขอบเขต
- โมเดลใช้ GPP เป็นตัวแปรอธิบายเพียงตัวเดียว ไม่ได้ควบคุมจำนวนประชากร ภารกิจภาครัฐ โครงสร้างเศรษฐกิจ หรือความต้องการบริการสาธารณะ
- แบบจำลองแสดงความสัมพันธ์ ไม่ได้พิสูจน์ว่างบประมาณทำให้ GPP เปลี่ยนแปลง หรือว่าค่าประมาณเป็นงบประมาณที่เหมาะสมที่สุด

## สรุปผล (Conclusion)

ค่าเฉลี่ย GPP และอัตราการเติบโตของ GPP แตกต่างกันระหว่างกลุ่มจังหวัดตามป้ายกำกับพรรคการเมืองในไฟล์ โดยกลุ่มก้าวไกลมีค่าเฉลี่ย GPP สูงที่สุด และกลุ่มประชาชาติมีค่าเฉลี่ยและค่ามัธยฐานการเติบโตสูงที่สุดที่ 5.66% ส่วนกลุ่มภูมิใจไทยมีค่าเฉลี่ยการเติบโตติดลบ ผลเหล่านี้เป็นการเปรียบเทียบเชิงพรรณนาเท่านั้น ไม่ได้แสดงว่าพรรคการเมืองเป็นสาเหตุให้ GPP สูงขึ้นหรือต่ำลง

## ข้อจำกัด (Limitations)

- จำนวนจังหวัดในแต่ละกลุ่มไม่เท่ากัน ตั้งแต่ 1 จังหวัดในกลุ่มไทยสร้างไทยถึง 23 จังหวัดในกลุ่มเพื่อไทย จึงควรระมัดระวังในการเปรียบเทียบค่าเฉลี่ย
- กลุ่มที่มีเพียง 1–2 จังหวัดอาจได้รับอิทธิพลจากข้อมูลของจังหวัดใดจังหวัดหนึ่งมาก และอาจไม่เป็นตัวแทนของกลุ่มได้ดี
- ค่าเฉลี่ย GPP อาจได้รับอิทธิพลจากจังหวัดที่มีเศรษฐกิจขนาดใหญ่ จึงควรพิจารณาค่ามัธยฐานควบคู่กัน
- วิธีการกำหนดพรรคการเมืองให้แต่ละจังหวัดยังเป็นข้อจำกัดสำคัญ หากไม่มีเกณฑ์การจับคู่ที่ชัดเจน ไม่ควรตีความป้ายกำกับว่าเป็นตัวแทนผลการดำเนินงานของพรรคโดยตรง
- การวิเคราะห์ไม่ได้ควบคุมปัจจัยอื่น เช่น โครงสร้างเศรษฐกิจ จำนวนประชากร และอุตสาหกรรมหลัก และไม่สามารถสรุปเหตุและผลได้
